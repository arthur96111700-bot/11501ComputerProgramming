問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

可以利用 2004 年是猴年作為基準，搭配 % 12（取餘數）來判斷生肖。若輸入的是民國年，則先加上 1911 轉換成西元年。

:::writing{variant="document" id="74126" title="Python 十二生肖判斷程式"}

year = int(input("請輸入年份（西元或民國）："))

# 如果輸入的是民國年，轉換成西元年
if year < 1912:
    year = year + 1911

animals = [
    "monkey", "rooster", "dog", "pig",
    "rat", "ox", "tiger", "rabbit",
    "dragon", "snake", "horse", "sheep"
]

# 2004 年是猴年
index = (year - 2004) % 12

print("生肖是：", animals[index])
:::

93

year = 93 + 1911

2004

生肖是： monkey

2025

animals[9] 是：

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙

Python 的一般 float 使用 IEEE 754 標準的雙精度浮點數（double precision）。

通常一個 Python float 會使用 64 bits 儲存：

┌────────┬───────────────┬────────────────────────────────────────┐
│ 1 bit  │    11 bits    │              52 bits                   │
│ 符號位 │    指數位      │              尾數位                    │
└────────┴───────────────┴────────────────────────────────────────┘
也就是：

1 bit：Sign（正負號）
11 bits：Exponent（指數）
52 bits：Fraction/Mantissa（有效位數的一部分）
因此，像 0.1 這種需要無限多個二進位位元才能表示的數字，Python 只能選擇一個最接近的有限二進位浮點數來儲存。

那要如何決定「最接近」哪一個？
IEEE 754 使用預設的四捨六入五成雙（round to nearest, ties to even）方式進行捨入。

簡單來說：

如果數值介於兩個可表示的浮點數之間，就選擇距離最近的；如果剛好在正中間，則選擇最低有效位為偶數的那一個。
因此 Python 並不是把 0.1 隨便截斷，而是：

真正的 0.1
    ↓
二進位無限循環
    ↓
IEEE 754 雙精度無法全部儲存
    ↓
選擇最接近的可表示數值
    ↓
儲存在電腦中
所以 Python 中的 0.1 實際上是「非常接近 0.1 的一個二進位浮點數」。

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？

執行：

print(0.1 + 0.2)
通常會得到：

0.30000000000000004
數學上當然是：

0.1 + 0.2 = 0.3
問題在於電腦儲存的 0.1 和 0.2 並不是完全精確的十進位 0.1 和 0.2。

大致可以想成：

0.1 → 最接近 0.1 的二進位浮點數
0.2 → 最接近 0.2 的二進位浮點數
兩個「近似值」相加後，結果可能會稍微大於真正的 0.3。

方法一：使用 round()
如果只是一般顯示，可以：

result = 0.1 + 0.2
print(round(result, 1))
輸出：

0.3
也可以比較：

print(round(0.1 + 0.2, 10))
輸出：

0.3
不過要注意：

round() 是把結果進行四捨五入，並沒有改變 float 本身有限精度的本質。
方法二：使用 math.isclose()
如果目的是「判斷兩個浮點數是否足夠接近」，更推薦：

import math

print(math.isclose(0.1 + 0.2, 0.3))
結果：

True
因為我們通常不需要要求兩個浮點數的二進位表示「完全相同」，而是判斷它們是否在可以接受的誤差範圍內。

方法三：需要精確十進位計算時使用 Decimal
例如金融計算：

from decimal import Decimal

a = Decimal("0.1")
b = Decimal("0.2")

print(a + b)
輸出：

0.3
注意：建立 Decimal 時最好使用字串：

Decimal("0.1")
而不是：

Decimal(0.1)
因為後者會先把 0.1 變成 Python 的二進位浮點數，才交給 Decimal

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。

不能一概而論。

科學計算
對許多科學、工程、統計計算而言，使用一般 float 是很正常的。

例如：

temperature = 23.5
distance = 10.2
speed = distance / 2.5
這類計算通常可以接受非常小的誤差。

科學計算更重要的是：

了解誤差有多大，以及誤差是否會在大量運算後累積。
因此不能看到：

0.30000000000000004
就認為 Python 算錯了。

它其實是受到浮點數有限精度的限制。

金融計算
金融則比較特殊。

例如：

price = 0.1
quantity = 3

print(price * quantity)
如果金額必須精確到小數點後兩位，直接使用二進位 float 並不是最理想的做法。

金融程式通常會考慮：

from decimal import Decimal

price = Decimal("0.10")
quantity = Decimal("3")

print(price * quantity)
結果：

0.30
或者，如果系統規定金額只能到「分」，也可以直接使用整數表示最小貨幣單位：

price = 10   # 10 分
quantity = 3

print(price * quantity)
結果：

30
也就是 30 分

電腦的 float 並不是「錯誤地計算 0.1 + 0.2」，而是因為 0.1 和 0.2 無法用有限的二進位浮點格式精確表示，所以運算結果會出現極小的表示誤差
