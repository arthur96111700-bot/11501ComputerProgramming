## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？
因為 Python 的 int 使用任意精度整數，會動態配置記憶體。

2. Python 的 `int` 是否具有固定的位元數限制？
沒有固定的位元數限制。

3. 什麼是 Arbitrary Precision（任意精度）？
任意精度是指整數大小可隨需要增加，不受固定 32 位或 64 位限制。

4. CPython 使用什麼物件來實作 `int`？
CPython 使用 PyLongObject 實作 int。

5. Python 整數大小真正的限制是什麼？
真正限制是可用的記憶體與系統資源。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？
因為 float 使用二進位浮點數，0.1、0.2 等十進位小數通常無法精確表示。

2. 浮點數是否能精確表示所有小數？
不能，有些小數只能用近似值表示。

3. 什麼是浮點數的精度（precision）？
精度（precision）：浮點數能表示數值的有效位數。

4. 什麼是浮點數的範圍（range）？
範圍（range）：浮點數能表示的最小到最大數值範圍。

5. 精度與範圍有何不同？
精度關心「多準確」；範圍關心「能表示多大或多小」

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：
1. AI 模型中的權重通常使用什麼型別儲存？ 如 FP32、FP16 或 BF16
2. 什麼是 FP64？ 64 位元浮點數，精度高、記憶體需求大。
3. 什麼是 FP32？ 32 位元浮點數，精度與效能取得平衡。
4. 什麼是 FP16？ 16 位元浮點數，節省記憶體、運算較快，但精度較低。
5. 什麼是 BF16？ 16 位元浮點數，範圍接近 FP32，但精度比 FP32 低，適合 AI 計算。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？ 因為 Python int 是任意精度整數，每個數值需要較多記憶體，且不適合大量矩陣的高速運算；AI 通常使用固定大小的浮點數更有效率

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？ 表示整數不受固定 32 或 64 位元限制，可依需要增加儲存空間。
2. 它是否代表可以儲存無限大的整數？ 不是無限大；實際仍受記憶體與系統資源限制。
3. CPython 如何實現這項特性？ CPython 使用 PyLongObject，將整數分成多個較小的數值單位儲存，需要時動態配置更多記憶體，因此可以表示非常大的整數。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？ int 使用任意精度，可動態增加記憶體，所以沒有固定的位元上限。
2. 為什麼 `float` 卻存在範圍限制？ float 使用固定大小（通常 64 位元）的浮點數格式，因此有最大、最小值限制。
3. 為什麼 `float` 可能產生誤差？ float 使用二進位表示，而許多十進位小數無法精確轉成二進位，只能用近似值表示，因此會產生誤差。

---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。 
Python int： 任意精度，重視數值正確性。
FP16／BF16： 固定精度，重視速度、記憶體與硬體效率。

2. 各自的優點與缺點。

方式	優點	缺點
int 任意精度	數值精確、可表示很大的整數	記憶體與運算成本可能增加
FP16／BF16	節省記憶體、運算速度快	精度較低，可能產生誤差或數值問題

3. 為什麼大型語言模型通常選擇較低精度的表示方式？
因為模型有大量權重與矩陣運算，使用 FP16／BF16 可以大幅降低記憶體需求，並利用 GPU 等硬體進行更高速的計算

4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？

如果全部改用 FP64 可能造成： 記憶體需求大幅增加 運算速度降低 硬體成本提高 同樣大小的模型可能需要更多 GPU／記憶體 因此 AI 通常會在精度、速度與記憶體使用量之間取得平衡

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

x = 10**1000 可以正常運作，是因為 Python 的 int 使用任意精度，會依需要增加記憶體來儲存超大的整數；但大型 AI 模型通常使用 FP32、FP16、BF16，因為它們是固定大小的數值格式，更適合大量矩陣運算。

記憶體使用量： FP16、BF16 只需要 16 位元，比任意精度整數或 FP64 更省記憶體。

計算速度： 固定大小的數值可以更快速地進行大量運算。

數值精度： FP16、BF16 雖然精度較低，但對 AI 訓練與推論通常已足夠。

GPU 運算效率： 現代 GPU 對 FP16、BF16 等低精度運算有專門硬體支援，因此能大幅提升矩陣運算效率。

簡單來說，Python int 重視「精確表示」；AI 模型則更重視「在可接受精度下，用更少的記憶體和更快的速度完成大量計算」。

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
