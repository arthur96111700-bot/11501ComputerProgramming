r = random.random()

print(f"假如 r = {r}")

binary = "0."
count = 0

while count < 20:
    x = x * 2

    if x >= 1:
        binary += "1"
        x -= 1
    else:
        binary += "0"

    count += 1

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

並查閱官方文件，了解：

- `bit_length()`
- `sys.getsizeof()`
- `PyLongObject`

def convert_group(num):
    result = ""
    zero_flag = False
    s = str(num).zfill(4)

    for i, ch in enumerate(s):
        d = int(ch)
        pos = 3 - i

        if d == 0:
            if result != "":
                zero_flag = True
        else:
            if zero_flag:
                result += "零"
                zero_flag = False

            result += digits[d] + small_units[pos]

    return result


def convert_number(n):
    if n == 0:
        return "零"

    groups = []
    while n > 0:
        groups.append(n % 10000)
        n //= 10000

    result = ""
    zero_between = False

    i = len(groups) - 1

    while i >= 0:
        group = groups[i]

        if group == 0:
            zero_between = True
        else:
            if result != "":
                if zero_between or group < 1000:
                    if not result.endswith("零"):
                        result += "零"

            result += convert_group(group) + big_units[i]
            zero_between = False

        i -= 1

    return result


n = int(input("請輸入正整數："))

習題三：請勿使用for、range，而是利用iterable、iter()、next()、while、StopIteration來設計九九乘法表。


i_iter = iter([1, 2, 3, 4, 5, 6, 7, 8, 9])

while True:
    try:
        i = next(i_iter)

        j_iter = iter([1, 2, 3, 4, 5, 6, 7, 8, 9])

        while True:
            try:
                j = next(j_iter)
                print(f"{i}x{j}={i*j:2}", end="\t")
            except StopIteration:
                break

        print()

    except StopIteration:
        break
